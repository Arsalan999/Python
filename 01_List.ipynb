# List 
## List is a Data type which can use for list data storage 
### this is mutable data type 
#### In Python lists can store heterogeneous data.


A list in Python is an ordered, mutable collection of items, defined using square brackets []. It can store elements of any data type and allows adding, removing, or changing items.

That means a single list can contain different data types—integers, strings, floats, objects, etc.

For Example :

```python
my_list = [1, "apple", 3.5]
```

## Some Function which use in List 

Here’s a clear and quick guide to **adding and removing elements in a Python list**:

---

# ✅ **Adding Elements**

### **1. `append()` — Add to the end**

```python
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)   # [1, 2, 3, 4]
```

### **2. `insert()` — Add at a specific position**

```python
my_list = [1, 2, 3]
my_list.insert(1, 10)   # insert 10 at index 1
print(my_list)          # [1, 10, 2, 3]
```

### **3. `extend()` — Add multiple elements**

```python
my_list = [1, 2, 3]
my_list.extend([4, 5])
print(my_list)   # [1, 2, 3, 4, 5]
```

### **4. Using `+` operator**

```python
my_list = [1, 2]
new_list = my_list + [3, 4]
print(new_list)  # [1, 2, 3, 4]
```

---

# ❌ **Removing Elements**

### **1. `remove()` — Remove by value**

(Removes the first matching element)

```python
my_list = [1, 2, 3, 2]
my_list.remove(2)
print(my_list)   # [1, 3, 2]
```

### **2. `pop()` — Remove by index (or last item by default)**

```python
my_list = [1, 2, 3]
item = my_list.pop(1)  # removes element at index 1
print(item)            # 2
print(my_list)         # [1, 3]
```

### **3. `del` — Delete by index or slice**

```python
my_list = [1, 2, 3]
del my_list[1]
print(my_list)   # [1, 3]
```

Delete a range:

```python
del my_list[0:2]
```

### **4. `clear()` — Remove all elements**

```python
my_list = [1, 2, 3]
my_list.clear()
print(my_list)   # []
```

12. Copying Lists (copy vs =)

Question:
Explain the difference between:

```
 a = [1,2,3]
b = a
c = a.copy()

```

Modify a and see which list changes.

````python
a = [1,2,3]
b = a
c = a.copy()


a.append(4)
print(a)
print (b)



print(c)
````

🔍 Explanation
b = a

b refers to the same list as a

Changing a → b changes too

Both share memory

c = a.copy()

Creates a new independent list

Changing a does NOT affect c
