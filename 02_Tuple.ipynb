# Tuple

A tuple in Python is an ordered, immutable collection of elements, created by placing items inside parentheses () and separating them with commas. Tuples are one of Python's built-in data types used to store collections of data. 

**Key Characteristics:**

**Ordered:** The items in a tuple have a defined order that does not change.

**Immutable:** Once a tuple is created, you cannot change, add, or remove items. Attempting to modify a tuple will result in a TypeError.

**Allows Duplicates:** Tuples can contain multiple items with the same value.

**Heterogeneous:** Tuple elements can be of any data type (strings, integers, floats, lists, etc.).

**Indexed:** Elements are accessed by a zero-based index, similar to lists. 

**Creating Tuples**
Tuples are typically created using parentheses (), but the comma-separated values are what actually define a tuple (this is known as "tuple packing").

4. Tuple Slicing

Question:
Given:

letters = ("a", "b", "c", "d", "e")


Print:

* First 3 items

* Last 2 items

* Items from index 1 to 3

```python
letters = ("a", "b", "c", "d", "e")

print("First 3 items :",letters[0:3:])

print("Last 2 items :",letters[3::])

print("Items from index 1 to 3 :",letters[1:4:])
```
